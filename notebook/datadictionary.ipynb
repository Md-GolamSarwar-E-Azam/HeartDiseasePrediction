
# Heart Disease Dataset: Data Dictionary

This dataset contains information about patients, including their medical history, blood pressure, cholesterol levels, ECG results, and other health-related measurements.

The following data dictionary explains each feature in simple, non-medical language.

## 1. Feature Descriptions

| Feature | Description | Example / Interpretation |
|---|---|---|
| `age` | Patient's age in years. | 55 means the patient is 55 years old. |
| `sex` | Patient's recorded gender. | 1 = Male, 0 = Female. |
| `cp` | Type of chest pain experienced by the patient. | Different values represent different types of chest pain. |
| `trestbps` | Blood pressure measured while the patient is resting. | 130 mmHg. |
| `chol` | Amount of cholesterol in the patient's blood. | 240 mg/dL. |
| `fbs` | Whether fasting blood sugar is greater than 120 mg/dL. | 1 = Yes, 0 = No. |
| `restecg` | Results of an ECG (electrocardiogram) test taken while the patient is resting. | Indicates whether the recorded heart electrical activity appears normal or shows certain abnormalities. |
| `thalach` | Maximum heart rate recorded during exercise testing. | 150 beats per minute. |
| `exang` | Whether the patient experienced chest pain during exercise. | 1 = Yes, 0 = No. |
| `oldpeak` | Amount of ST-segment depression observed during exercise compared with rest. | Higher values indicate greater ST depression. |
| `slope` | Direction or pattern of the ST segment during peak exercise. | Describes whether the pattern is upward, flat, or downward. |
| `ca` | Number of major blood vessels visible through a heart imaging test. | Values generally range from 0 to 3. |
| `thal` | Result of a thallium-based heart stress test, which helps assess blood flow to the heart muscle. | Different values represent different test findings. |
| `target` | Indicates whether heart disease is present according to the dataset's classification. | 1 = Disease present, 0 = No disease. |

## 2. Understanding Categorical Features

Some features contain numerical values that represent categories rather than actual measurements.

### Chest Pain (`cp`)

Describes the type of chest pain experienced by the patient.

| Value | Medical Term | Simple Explanation |
|---|---|---|
| 0 | Typical angina | Chest pain following the usual pattern associated with reduced blood supply to the heart. |
| 1 | Atypical angina | Chest pain that has some, but not all, typical characteristics. |
| 2 | Non-anginal pain | Chest pain that does not follow the typical angina pattern. |
| 3 | Asymptomatic | No chest pain symptoms. |

### Resting ECG (`restecg`)

Describes the findings from an ECG test performed while the patient is resting.

| Value | Medical Term | Simple Explanation |
|---|---|---|
| 0 | Normal | No significant abnormality reported. |
| 1 | ST-T wave abnormality | Changes in the ECG pattern that may indicate a heart-related issue. |
| 2 | Left ventricular hypertrophy | Signs of increased muscle thickness in the heart's main pumping chamber. |

### Exercise ECG Slope (`slope`)

Describes the direction of the ST segment during peak exercise.

| Value | Medical Term | Simple Explanation |
|---|---|---|
| 0 | Upsloping | The ECG pattern moves upward. |
| 1 | Flat | The ECG pattern remains relatively flat. |
| 2 | Downsloping | The ECG pattern moves downward. |

### Thallium Stress Test (`thal`)

Describes findings from a test used to assess blood flow to the heart muscle.

| Value | Medical Term | Simple Explanation |
|---|---|---|
| 0 | Unknown | Test result not specified. |
| 1 | Normal | Normal blood-flow pattern. |
| 2 | Fixed defect | An area shows persistently reduced tracer uptake, potentially indicating previous heart muscle damage. |
| 3 | Reversible defect | An area shows reduced tracer uptake during exercise that improves at rest. |

## 3. Understanding Important Medical Measurements

A few features deserve additional attention because their names may not be immediately understandable.

- **`trestbps` (Resting Blood Pressure):** Measures blood pressure while the patient is resting. It is generally recorded in mmHg.
- **`chol` (Cholesterol):** Measures the amount of cholesterol in the blood, generally in mg/dL.
- **`thalach` (Maximum Heart Rate):** Records the highest heart rate achieved during exercise testing, measured in beats per minute.
- **`oldpeak` (ST Depression):** Measures the amount of ST-segment depression during exercise compared with rest. Higher values indicate greater ST depression.

## 4. Target Variable

The `target` column is the variable we want to predict using machine learning.

| Value | Meaning |
|---|---|
| 0 | No heart disease |
| 1 | Heart disease present |

Our machine learning objective is to predict whether a patient belongs to the disease or no-disease category based on the available features.
